## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
    """Return the upper-cased read, whether it is valid, and any errors found."""
    # Write your code here.

    sequence = sequence.upper()   #this makes the bases uppercase so that bases that were lowercase are not rejected
    errors = []

    if len(sequence) != len(reference):  #this checks length first so "wrong_length" always comes before "invalid_characters"
      errors.append("wrong_length")

    for base in sequence:
      if base not in "ACGT":
        errors.append("invalid_characters")
        break

    return {"sequence": sequence,        #a read is only valid if there are no errors
            "valid": len(errors) == 0,
            "errors": errors}
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """Return a list of mismatches (1-based position, reference base, read base)."""
    # Write your code here.
    mutations = []

    for i in range(len(reference)):
      if sequence[i] != reference[i]:
        mutations.append({"position": i + 1,               #python will index from 0, position starts at 1
                          "reference_base": reference[i],
                          "read_base": sequence[i],
                          })

    return mutations
````

### 💣 Test your code

````python
# Test your function with this cell

# This is an example -- modify to test your program more
test_mutations = find_mutations("GTGA", "ATGC")
assert test_mutations == [{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]

#read matches the reference so there are no mutations
assert find_mutations("ATGC", "ATGC") == []

#only the last base is different so the position should be 4
assert find_mutations("ATGA", "ATGC") == [{"position": 4,
                                           "reference_base": "C",
                                           "read_base": "A"}]

# Print the function's output
print(test_mutations)
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """Return a new list of reads annotated with status, errors and mutations"""
    # Write your code here.

    analysed_reads = []

    for read in reads:
        validation = validate_read(read["sequence"], reference)   #check the read first because find_mutations only works on valid reads

        if validation["valid"]:
            status = "valid"
            mutations = find_mutations(validation["sequence"], reference)   #use the upper case sequence so lowercase reads don't give false mismatches
        else:
            status = "invalid"   #invalid reads can't be compared as the wrong length would stop the code so they return an empty list
            mutations = []

        analysed_reads.append({               #i made a new dictionary so the original reads are not changed
            "read_id": read["read_id"],
            "sample": read["sample"],
            "group": read["group"],
            "sequence": validation["sequence"],
            "status": status,
            "errors": validation["errors"],
            "mutations": mutations,
            })

    return analysed_reads
````

````python
# Write your code here.
def add_count(counts, key):
  """Adds 1 to the count for key, starting from 1 if it is new"""
  if key in counts:
    counts[key] = counts[key] + 1
  else:
    counts[key] = 1

def find_positions(position_counts, max_positions=3):
  """Return up to 3 positions with the most mismatches"""
  #sort by highest count, then by lowest position
  remaining = sorted(position_counts)
  top_positions = []
  for i in range(min(max_positions, len(remaining))):
    best = remaining[0]
    for position in remaining:
      if position_counts[position] > position_counts[best]:
        best = position
    top_positions.append(best)
    remaining.remove(best)
  return top_positions


def count_mismatches(sample_reads):
  """Return the mismatch summary for all analysed reads from one sample"""
  valid_reads = 0
  invalid_reads = 0
  mismatch_reads = 0
  mismatch_events = 0
  substitution_counts = {}
  position_counts = {}

  for read in sample_reads:
    if read["status"] != "valid":
      invalid_reads += 1
    else:
      valid_reads += 1
      if len(read["mutations"]) > 0:
        mismatch_reads += 1

      for mutation in read["mutations"]:
          mismatch_events += 1
          substitution = mutation["reference_base"] + ">" + mutation["read_base"]
          add_count(substitution_counts, substitution)
          add_count(position_counts, mutation["position"])

  if valid_reads == 0:
     mismatch_percentage = None

  else:
     mismatch_percentage = mismatch_reads / valid_reads * 100

  return{"group": sample_reads[0]["group"],
        "valid_reads": valid_reads,
        "invalid_reads": invalid_reads,
        "mismatch_reads": mismatch_reads,
        "mismatch_percentage": mismatch_percentage,
        "mismatch_events": mismatch_events,
        "substitution_counts": substitution_counts,
        "position_counts": position_counts,
        "top_positions": find_positions(position_counts)
        }

def print_tables(sample_summary):
  """Print a results table and a substitution table for each group"""
  group_names = sorted({sample_stats["group"]
                        for sample_stats in sample_summary.values()})
  for group_name in group_names:
    sample_names = sorted(name for name, stats_for_sample in sample_summary.items()
        if stats_for_sample["group"] == group_name)

    print(f"\n===== Group: {group_name} =====")

#table 1 = one row per sample
    print(f'{"Sample":<10}{"Valid":>8}{"Invalid":>9}{"% mismatch":>12}{"Events":>8}   Top positions')
    print("-" * 62)
    for sample_name in sample_names:
          sample_stats = sample_summary[sample_name]
          if sample_stats["mismatch_percentage"] is None:
             percentage_text = "n/a"
          else:
             percentage_text = f'{sample_stats["mismatch_percentage"]:.2f}'
          positions_text = ", ".join(str(p) for p in sample_stats["top_positions"]) or "none"
          print(f'{sample_name:<10}{sample_stats["valid_reads"]:>8}{sample_stats["invalid_reads"]:>9}'
                f'{percentage_text:>12}{sample_stats["mismatch_events"]:>8}   {positions_text}')

   #table 2 = one row per substitution and one column per sample
    substitution_types = sorted({substitution
                                 for sample_name in sample_names
                                 for substitution in sample_summary[sample_name]["substitution_counts"]})
    print(f"\nSubstitutions in {group_name} (count and % of valid reads)")
    print(f'{"Substitution": <14}' + "".join(f'{sample_name:>16}' for sample_name in sample_names))
    print("-" * (14  + 16 * len(sample_names)))
    for substitution in substitution_types:
      table_row = f"{substitution:<14}"
      for sample_name in sample_names:
        sample_stats = sample_summary[sample_name]
        count = sample_stats["substitution_counts"].get(substitution, 0)
        if sample_stats["valid_reads"] == 0:
                    cell_text = f"{count} (n/a)"
        else:
          cell_text = f'{count} ({count / sample_stats["valid_reads"] * 100:.2f}%)'
        table_row += f"{cell_text:>16}"
      print(table_row)


def summarise_samples(analysed_reads):
    """Summarise mismatches for each sample, print the tables and return the results."""
    reads_per_sample = {}
    for read in analysed_reads:
        if read["sample"] not in reads_per_sample:
            reads_per_sample[read["sample"]] = []
        reads_per_sample[read["sample"]].append(read)

    sample_summary = {}
    for sample_name, sample_reads in reads_per_sample.items():
        sample_summary[sample_name] = count_mismatches(sample_reads)

    print_tables(sample_summary)
    return sample_summary

analysed_reads = analyse_reads(reads, reference)
sample_summary = summarise_samples(analysed_reads)
````

## Step 5 — Exploratory comparison of the top mismatch positions

## Function:
`compare_positions(sample_summary, control_group="control", treated_group="treatment")`

### 👍 You can assume

- `control_group` and `treated_group` are strings

### 🎯 Aims

- Using the `position_counts` returned by `summarise_samples`, sum the mismatch counts at every position, **combining samples from `control_group` and `treatment_group`** (you should end up with a 1D array/list with the same length as a read).
- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.
- For each position selected, calculate **mismatch percentages** separately for each biological sample.
- Compare treatment groups using the supplied statistical function `calculate_statistics` (see below).
- **Print out** the results of the statistical comparison clearly in standard scientific style, including appropriate statistical details based on the test used (a two-sided Welch's t-test).

🛠 **Parameters:**

- `sample_summary`: output from `summarise_samples` function.
- `control_group` and `treated_group`: labels identifying the groups to compare.

### 🧮 Sample-level calculations

For each selected position and sample, calculate:

**Mismatch percentage = valid reads with a mismatch at that position ÷ total valid reads × 100**

Count any substitution at the selected position. Include samples with zero mismatches, but exclude samples with no valid reads and report their exclusion.

Reuse helper functions from Step 4 where appropriate to reduce code complexity and repetition.

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

### 📋 Return format

Return a list with one dictionary per selected position:

```python
[
    {
        "position": 42,
        "control_n": 3,
        "treated_n": 3,
        "control_mean": 0.5,
        "treated_mean": 2.0,
        "difference": 1.5,
        "statistic": ...,
        "p_value": ...
    }
]
```

Means are sample mismatch percentages; `difference` is **treated mean − control mean**, in percentage points. Use `None` for results that cannot be calculated.

Print **one brief line per position**, reporting group means, their difference, sample numbers and p-value.

### 💻 Enter your code here

````python
# Write your code here.
def calculate_mean(values):
  """return the mean of a list of numbers, or None if the list is empty"""
  if len(values) == 0:
    return None
  total = 0
  for value in values:
    total = total + value
  return total / len(values)

def calculate_variance(values):
  """return the sample variance of a list of at least 2 numbers"""
  mean = calculate_mean(values)
  squared_total = 0
  for value in values:
    squared_total = squared_total + (value - mean) ** 2
  return squared_total / (len(values) - 1)

def calculate_df(control_percentages, treated_percentages):
  """return the degrees of freedom for Welches t-test"""
  control_var = calculate_variance(control_percentages) / len(control_percentages)
  treated_var = calculate_variance(treated_percentages) / len(treated_percentages)
  top = (control_var + treated_var) ** 2
  bottom = (control_var ** 2 / (len(control_percentages) - 1) + treated_var ** 2 / (len(treated_percentages) - 1))
  return top / bottom

def calculate_percentages(sample_summary, group_name, position):
  """return the mismatch percentage at one position for each sample in a group"""
  percentages = []
  for sample_stats in sample_summary.values():
    if sample_stats["group"] == group_name and sample_stats["valid_reads"] > 0:
      count = sample_stats["position_counts"].get(position, 0)
      percentage = count / sample_stats["valid_reads"] * 100
      percentages.append(percentage)
  return percentages

def format_number(value):
  if value is None:
    return "n/a"
  return f"{value:.2f}"

def format_test(result, control_percentages, treatment_percentages):
  if result["p_value"] is None:
    return "Welch's t-test not possible"
  df = calculate_df(control_percentages, treatment_percentages)
  if result["p_value"] < 0.001:
    p_value = "p < 0.001"
  else:
    p_value = f"p = {result["p_value"]:.3f}"
  return f"Welch t({df:.1f}) = {result["statistic"]:.2f}, {p_value}"

def compare_positions(sample_summary, control_group="control", treated_group="treatment"):
  combined_counts = {}
  excluded_samples = []
  for sample_name, sample_stats in sample_summary.items():
    if sample_stats["group"] == control_group or sample_stats["group"] == treated_group:
      if sample_stats["valid_reads"] == 0:
        excluded_samples.append(sample_name)

      for position, count in sample_stats["position_counts"].items():
        combined_counts[position] = combined_counts.get(position, 0) + count
  if len(excluded_samples) > 0:
    print("Excluded samples:", ", ".join(excluded_samples))
  if len(combined_counts) == 0:
    print("No mismatches found")
    return []
  comparison_results = []
  for position in find_positions(combined_counts):
    control_percentages = calculate_percentages(sample_summary, control_group, position)
    treated_percentages = calculate_percentages(sample_summary, treated_group, position)

    statistical_result = calculate_statistics(control_percentages, treated_percentages)

    control_mean = calculate_mean(control_percentages)
    treated_mean = calculate_mean(treated_percentages)

    if control_mean is None or treated_mean is None:
      difference = None
    else:
      difference = treated_mean - control_mean

    comparison_results.append({
        "position": position,
        "control_n": len(control_percentages),
        "treated_n": len(treated_percentages),
        "control_mean": control_mean,
        "treated_mean": treated_mean,
        "difference": difference,
        "statistic": statistical_result["statistic"],
        "p_value": statistical_result["p_value"]
    })

    print(
        f"Position {position}: "
        f"{control_group} {format_number(control_mean)}% (n={len(control_percentages)}), "
        f"{treated_group} {format_number(treated_mean)}% (n={len(treated_percentages)}), "
        f"difference {format_number(difference)} percentage points, "
        f"{format_test(statistical_result, control_percentages, treated_percentages)}"
    )

  return comparison_results
````
