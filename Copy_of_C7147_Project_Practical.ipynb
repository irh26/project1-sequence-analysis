# 1️⃣ Code style and function design (≈ 8 minutes)

### 🔎 Why organise code into functions?
A good function usually does **one clearly defined job**, takes the information it needs as **parameters**, and **returns** a useful result. Prefer returning data to printing inside a function when another function might need to use the result.

Good style helps other researchers understand your analysis:
- Use descriptive `snake_case` names, e.g. `count_gc_bases` rather than `f`.
- Use four spaces for indentation and space around operators.
- Write a short docstring to explain the function's purpose, inputs and output.
- Use comments for *why* something is done; avoid comments that merely repeat the code.
- Avoid repeating nearly identical blocks of code. Extract a helper function when it makes the code clearer.

**Useful rule:** only validate inputs that you are **not allowed to assume** are valid.

### 👀 Worked example — from unclear code to a reusable function

Both snippets below count the number of G and C bases in a sequence. Which version is easier to use in a larger analysis?

```python
# Less reusable: uses a fixed value and only prints
s = "ATGCC"
n = 0
for x in s:
    if x == "G" or x == "C":
        n += 1
print(n)
```

```python
def count_gc_bases(sequence):
    """Return the number of G and C bases in a DNA sequence."""
    gc_count = 0
    for base in sequence:
        if base in "GC":
            gc_count += 1
    return gc_count

print(count_gc_bases("ATGCC"))
print(count_gc_bases("AAAA"))
```

# 3️⃣ Displaying results in a table with f-strings (≈ 4 minutes)

Clear output is part of writing understandable scientific code. You can make a simple **text table** using `print()` and **f-strings**, without importing a library.

Inside an f-string, the part after `:` controls formatting:

| Format | Meaning | Example |
|---|---|---|
| `{value:<12}` | Left-align in a field 12 characters wide | Names and labels |
| `{value:>8}` | Right-align in a field 8 characters wide | Numbers |
| `{value:>10.2f}` | Right-align a number in 10 spaces, with 2 decimal places | Measurements |

The field width controls **spacing**, not the value being stored. A header, separator, and loop make the table readable.

### 👀 Worked example — seedling growth

Imagine measuring the heights of seedlings on different days. This is **separate from Project 1's sequencing summaries**. Run the cell and notice how the columns line up.

```python
measurements = [
    {"day": 1, "height_cm": 1.25},
    {"day": 3, "height_cm": 2.8},
    {"day": 5, "height_cm": 4.375},
]

print(f'{"Day":<8}{"Height (cm)":>14}')
print("-" * 22)
for measurement in measurements:
    print(f'{measurement["day"]:<8}{measurement["height_cm"]:>14.2f}')
```
